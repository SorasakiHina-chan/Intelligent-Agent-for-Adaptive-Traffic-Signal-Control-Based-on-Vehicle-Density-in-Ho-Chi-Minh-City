```python
import sys
from IPython.display import HTML, display

sys.path.insert(0, "/content/Traffic_Agent_Project/source_code")

import matplotlib.pyplot as plt
from visualization import build_demo_frames, animate_intersection

frames = build_demo_frames(
    scenario="high",
    method="adaptive"
)

fig, anim, update = animate_intersection(
    frames,
    title="AI ADAPTIVE TRAFFIC LIGHT",
    speed=5,
    show=False
)

plt.close(fig)

display(HTML(anim.to_jshtml()))
```

```python
frames_fixed = build_demo_frames(
    scenario="high",
    method="fixed"
)

fig2, anim2, update2 = animate_intersection(
    frames_fixed,
    title="FIXED TRAFFIC LIGHT",
    speed=5,
    show=False
)

plt.close(fig2)

display(HTML(anim2.to_jshtml()))
```
